## 2 · Which model is this code running?

Read the chunk. **Before running: how many numbers does `spreads` hold when the loop finishes, and is each one built from 100 or from 1,000 draws?**

```python
import numpy as np

rng = np.random.default_rng(47)
spreads = []
for depth in (10, 100, 1000):
    fractions = []
    for _ in range(200):
        heads = 0
        for _ in range(depth):
            if rng.random() < 0.5:
                heads += 1
        fractions.append(heads / depth)
    spreads.append(np.std(fractions))
print(np.round(spreads, 3))
```

`spreads` will contain three numbers, one per depth, each summarizing 200 runs of the model at depth flips. There are three nested loops here, each doing a different job: the outer loop varies `depth`, the middle loop repeats the simulation 200 times at that `depth`, and the inner loop runs the model itself (the individual flips). The goal of this exercise is identifying which loop does which job.
